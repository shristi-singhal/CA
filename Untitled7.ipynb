```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("fahmidachowdhury/customer-segmentation-data-for-marketing-analysis")

print("Path to dataset files:", path)
```

```python
import os

print("Files in dataset folder:")
print(os.listdir(path))
```

```python
import pandas as pd
import os

# Find CSV files automatically
csv_files = [file for file in os.listdir(path) if file.endswith(".csv")]

print("CSV files:", csv_files)

# Load the first CSV file
file_path = os.path.join(path, csv_files[0])

df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)

df.head()
```

```python
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe(include="all")
```

```python
import numpy as np

# Check missing values
print(df.isnull().sum())

# Remove duplicates
df = df.drop_duplicates()

# Fill missing numerical values with median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print("Dataset shape after cleaning:", df.shape)
```

```python
df.head()
```

```python


rfm = df[[
    'membership_years',
    'purchase_frequency',
    'last_purchase_amount',
    'spending_score'
]].copy()


rfm.columns = [
    'Membership_Years',
    'Frequency',
    'Monetary',
    'Spending_Score'
]

rfm.head()
```

```python
print(rfm.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm)

print("Scaled data shape:", rfm_scaled.shape)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_scaled)

    inertia.append(kmeans.inertia_)


plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(rfm_scaled)

    score = silhouette_score(
        rfm_scaled,
        labels
    )

    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.show()
```

```python
optimal_k = 4

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

rfm['KMeans_Cluster'] = kmeans.fit_predict(
    rfm_scaled
)

print(rfm['KMeans_Cluster'].value_counts())

rfm.head()
```

```python
kmeans_score = silhouette_score(
    rfm_scaled,
    rfm['KMeans_Cluster']
)

print("K-Means Silhouette Score:", kmeans_score)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

pca_result = pca.fit_transform(rfm_scaled)

rfm['PCA1'] = pca_result[:, 0]
rfm['PCA2'] = pca_result[:, 1]

print(
    "Explained Variance Ratio:",
    pca.explained_variance_ratio_
)
```

```python
import seaborn as sns

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=rfm,
    x='PCA1',
    y='PCA2',
    hue='KMeans_Cluster',
    palette='Set2',
    s=70
)

plt.title("K-Means Customer Segmentation using PCA")

plt.show()
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42,
    max_iter=1000
)

tsne_result = tsne.fit_transform(rfm_scaled)

rfm['TSNE1'] = tsne_result[:, 0]
rfm['TSNE2'] = tsne_result[:, 1]

rfm.head()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=rfm,
    x='TSNE1',
    y='TSNE2',
    hue='KMeans_Cluster',
    palette='Set2',
    s=70
)

plt.title("K-Means Customer Segmentation using t-SNE")

plt.show()
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.8,
    min_samples=5
)

rfm['DBSCAN_Cluster'] = dbscan.fit_predict(
    rfm_scaled
)

print(
    rfm['DBSCAN_Cluster']
    .value_counts()
    .sort_index()
)
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=rfm,
    x='PCA1',
    y='PCA2',
    hue='DBSCAN_Cluster',
    palette='Set1',
    s=70
)

plt.title("DBSCAN Customer Segmentation using PCA")

plt.show()
```

```python
cluster_profile = rfm.groupby(
    'KMeans_Cluster'
).agg(

    Customers=('KMeans_Cluster', 'count'),

    Avg_Membership_Years=('Membership_Years', 'mean'),

    Avg_Frequency=('Frequency', 'mean'),

    Avg_Monetary=('Monetary', 'mean'),

    Avg_Spending_Score=('Spending_Score', 'mean')

).round(2)

cluster_profile
```

```python
def assign_persona(row):

    if (
        row['Avg_Frequency'] >= cluster_profile['Avg_Frequency'].median()
        and row['Avg_Monetary'] >= cluster_profile['Avg_Monetary'].median()
        and row['Avg_Spending_Score'] >= cluster_profile['Avg_Spending_Score'].median()
    ):
        return "Champions / High-Value Customers"

    elif row['Avg_Monetary'] >= cluster_profile['Avg_Monetary'].median():
        return "High Spending Customers"

    elif row['Avg_Frequency'] < cluster_profile['Avg_Frequency'].median():
        return "Occasional Customers"

    else:
        return "Regular Customers"


cluster_profile['Persona'] = cluster_profile.apply(
    assign_persona,
    axis=1
)

cluster_profile
```

```python
marketing_strategy = {

    "Champions / High-Value Customers":
    "Provide VIP rewards, exclusive discounts and early access to products.",

    "High Spending Customers":
    "Use premium product recommendations and personalized offers.",

    "Occasional Customers":
    "Send discount coupons and reminder campaigns to increase purchase frequency.",

    "Regular Customers":
    "Use loyalty programs and cross-selling recommendations."
}


cluster_profile['Marketing_Strategy'] = (
    cluster_profile['Persona']
    .map(marketing_strategy)
)

cluster_profile
```

```python
rfm.to_csv(
    "customer_segmentation_results.csv",
    index=False
)

cluster_profile.to_csv(
    "customer_marketing_personas.csv"
)

print("Customer segmentation completed successfully!")
```
